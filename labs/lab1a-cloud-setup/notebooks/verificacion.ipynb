```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

# getOrCreate() funciona tanto en EMR (donde ya existe una sesiÃ³n activa
# administrada por el clÃºster) como en un entorno local con pyspark
# instalado, sin necesitar ramas de cÃ³digo distintas.
spark = SparkSession.builder.appName("ST1630-Lab1a-Verificacion").getOrCreate()

# EDITAR: reemplaza por el bucket que creaste en setup_s3.sh
# (convenciÃ³n: st1630-{tu-usuario}-{aÃ±o})
BUCKET = "st1630-efcortesr-2026"

# En EMR, S3 se referencia directamente con el esquema s3://.
# En local (con las credenciales de AWS Academy exportadas), la misma
# ruta tambiÃ©n funciona porque Spark usa el conector S3A por debajo.
ruta_parquet = f"s3://{BUCKET}/bronze/ventas/prueba_parquet.parquet"
ruta_csv = f"s3://{BUCKET}/bronze/ventas/prueba_csv.csv"

df_parquet = spark.read.parquet(ruta_parquet)

df_parquet.printSchema()
df_parquet.show(5, truncate=False)

# Si ves el schema y las filas de arriba, tu datalake funciona
# correctamente de punta a punta: bucket, permisos IAM y clÃºster EMR.
print("Filas leÃ­das:", df_parquet.count())
```

```python
import time

df_csv = spark.read.option("header", "true").option("inferSchema", "true").csv(ruta_csv)

# Misma consulta sobre ambos formatos: filtrar por regiÃ³n y categorÃ­a,
# y agregar el total vendido -- el tipo de consulta selectiva que se
# beneficia de predicado pushdown y column pruning en formatos
# columnares (visto en la clase de S4, slide de Parquet vs. CSV).

def benchmark(df, nombre):
    inicio = time.time()
    resultado = (
        df.filter((F.col("region") == "BogotÃ¡") & (F.col("categoria") == "ElectrÃ³nica"))
          .groupBy("producto")
          .agg(F.sum("total").alias("total_vendido"))
          .orderBy(F.col("total_vendido").desc())
          .collect()  # acciÃ³n -- fuerza la ejecuciÃ³n real, no solo el plan
    )
    duracion = time.time() - inicio
    print(f"{nombre}: {duracion:.3f} s ({len(resultado)} filas de resultado)")
    return duracion

tiempo_parquet = benchmark(df_parquet, "Parquet")
tiempo_csv = benchmark(df_csv, "CSV")

ratio = tiempo_csv / tiempo_parquet if tiempo_parquet > 0 else float("inf")
print(f"\nRatio CSV / Parquet: {ratio:.2f}x")
```

## Analisis - resultados de la verificacion

### a) Tamano en disco

`prueba_parquet.parquet` peso aproximadamente **185.4 KB**, mientras que `prueba_csv.csv` peso aproximadamente **798.3 KB**. En otras palabras, el CSV quedo alrededor de **4.30 veces mas grande** que el Parquet. Esto nos confirma en pequeno lo que esperabamos de un formato columnar: Parquet guarda los datos de forma mas compacta y aprovecha mejor la compresion, especialmente cuando hay columnas repetidas como `region`, `categoria` y `canal`.

### b) Tiempo de la consulta

En nuestra ejecucion dentro de EMR Studio, la consulta sobre Parquet tardo **0.993 s** y la consulta sobre CSV tardo **0.716 s**. Tambien vimos jobs intermedios completados correctamente en Spark UI, por ejemplo lecturas/collects con tiempos cercanos a **2.303 s**, **0.537 s**, **0.528 s** y una lectura CSV de **0.46 s**. Lo importante es que ambos formatos se pudieron leer desde S3 usando el cluster EMR, y Spark conto **10.000 filas**, lo que valida el flujo completo bucket-IAM-EMR-Studio.

### c) Ratio de mejora

El ratio obtenido fue **CSV / Parquet = 0.72x**. En esta corrida no vimos una mejora de Parquet frente a CSV; de hecho, CSV salio un poco mas rapido. No lo interpretamos como que CSV sea mejor en general, sino como efecto del tamano pequeno del dataset, el overhead fijo de Spark, el arranque de la sesion, la inferencia/lectura en un cluster minimo y posibles efectos de cache. Con solo 10.000 filas, el costo de planear y ejecutar el job puede pesar mas que la ventaja del formato columnar. En datasets mas grandes o consultas que lean pocas columnas, esperamos que Parquet muestre mejor su ventaja.

### d) Conexion con el Teorema CAP

S3 se puede entender como una decision cercana a **CP** porque prioriza consistencia y tolerancia a particiones: despues de una escritura exitosa, las lecturas deben ver el objeto correcto y no una version vieja. Para lograr esa garantia, S3 replica los datos entre multiples zonas de disponibilidad antes de considerar segura la operacion. El costo de esa decision es que, ante fallas o particiones de red, el sistema puede preferir rechazar o demorar una operacion antes que responder con datos inconsistentes. Para nuestro datalake esto es importante porque Spark asume que los archivos que lee en Bronze son la version correcta de la fuente.
